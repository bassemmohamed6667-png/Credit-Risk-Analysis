```python
#2 discovering the missing value
df.isna().sum()
```

```python
#3 Checking Duplicates
df.duplicated().sum()
```

```python
#4 Handling Outliers
df.loc[df['person_age'] > 80, 'person_age'] = pd.NA
```

```python
df.loc[df['person_emp_length'] > 80, 'person_emp_length'] = pd.NA
```

```python
#5 Checking the results
df['person_age'].isna().sum(),
```

```python
df['person_emp_length'].isna().sum()
```

```python
#6 Handling the missing values
df['person_age'] = df['person_age'].fillna(df['person_age'].median())
```

```python
df['person_emp_length'] = df['person_emp_length'].fillna(df['person_emp_length'].median())
```

```python
df['loan_int_rate'] = df['loan_int_rate'].fillna(df['loan_int_rate'].mean())
```

```python
#7 final checking
df.isna().sum()
```

```python
# Save the cleaned data as Excel
df.to_excel("cleaned_credit_risk_dataset.xlsx", index=False)
```
